```python
import gtfs_kit

print("Importing GTFS feed...")
feed = gtfs_kit.read_feed("itm_all_gtfs.zip", dist_units="km")
print(f"GTFS feed contains {len(feed.agency)} agencies, {len(feed.routes)} routes, and {len(feed.trips)} trips.")
```

```python
routes = feed.get_routes()
feed.map_routes(route_ids=routes["route_id"].tolist())
```
